```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

n = 5
p = 3

X_matriz = np.array([
    [25, 3000, 8.5],
    [45, 8000, 4.2],
    [22, 1500, 9.8],
    [35, 5000, 6.1],
    [28, 4200, 7.5]
])

df_X = pd.DataFrame(X_matriz, columns=['Edad', 'Salario', 'Engagement'], 
                    index=['U1', 'U2', 'U3', 'U4', 'U5'])

print("Nuestra Matriz Multidimensional X (n=5, p=3):")
display(df_X)
```

```python
vector_u2 = X_matriz[1] 
vector_u4 = X_matriz[3]

print(f"Coordenadas Usuario 2: {vector_u2}")
print(f"Coordenadas Usuario 4: {vector_u4}")

distancia = np.linalg.norm(vector_u2 - vector_u4)

print(f"\nLa distancia geométrica exacta entre estos dos usuarios es: {distancia:.2f} unidades.")
```

```python
X_matriz = (X_matriz - X_matriz.mean()) / X_matriz.std()
```

```python
vector_u2 = X_matriz[1] 
vector_u4 = X_matriz[3]

print(f"Coordenadas Usuario 2: {vector_u2}")
print(f"Coordenadas Usuario 4: {vector_u4}")

distancia = np.linalg.norm(vector_u2 - vector_u4)

print(f"\nLa distancia geométrica exacta entre estos dos usuarios es: {distancia:.2f} unidades.")
```

#### Explica cómo cambió la distancia y por qué la segunda métrica es la única que un modelo de Inteligencia Artificial debería usar.

##### Cambió de forma que antes de hacer la estandarización, la distancia era por decirlo así totalmente la columna 2, entonces el resto de columnas se convierten en insignificantes, y al hacer la estadarización todos los rangos a dsitancias de entre -1 y 3.

##### Los modelos de inteligencia artificial no entienden el significado semántico ni las unidades físicas de las columnas, a estos simplemente les interesan y procesan las magnitudes magnitudes numéricas, entonces es por eso que la eliminación de sesgo es tan importante al igual que la normalización o estandarización.
