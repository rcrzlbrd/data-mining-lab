### Pregunta A: 
#### Observen la columna Fare (Tarifa del boleto). ¿Cuál es la diferencia matemática entre el percentil 75% (el tercer cuartil) y el valor Máximo (max)? Como ingenieros de software, ¿qué les dice esta enorme dispersión estadística sobre la homogeneidad de los precios que se cobraron en el barco?

```python
print(df['Fare'].max() - df['Fare'].quantile(0.75))
```

##### Que un gran porcerntaje de los pasajeros, en este caso el 75%, pagó 31 dólares o menos por los boletos, mientras que el último 25% pagó cifras mucho más elevadas. Lo que puede decir que la media no refleja la realidad por esta dispersión.
