```python
import pandas as pd
 
url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/titanic.csv"
df = pd.read_csv(url)
```

### Asimetría (Skewness): 
#### Investiguen cómo aplicar la función .skew() en la columna Age. Si el resultado es positivo o negativo, ¿qué significa físicamente sobre la edad de los pasajeros en el barco? (¿Había más jóvenes o más ancianos?).

```python
age_skewness = df['Age'].skew()

print(age_skewness)
```

#### En este caso tuvimos una asimetría positiva indica que la cola de la distribución se alarga hacia la derecha que quiere decir valores altos. Por lo tanto, la masa principal de datos está concentrada en el extremo izquierdo. Entonces con esto podemos decir que había más personas jóvenes porque la cola de distribución al estar en el lado derecho quiere decir que había unas pocas personas ancianas pero que en su mayoría eran jóvenes.


### Curtosis (Kurtosis):
#### Investiguen cómo aplicar la función .kurt() en la columna Fare. ¿Qué significa este número sobre el "grosor" de las colas de la distribución? ¿Confirma esto la existencia de los outliers que encontramos hoy en clase?

```python
fare_kurtosis = df['Fare'].kurt()

print(fare_kurtosis)
```

##### El grosor es la probabilidad o masa de datos hay en los extremos lejanos, y si esto confirma la existencia de outliers.

### Tablas Cruzadas (Cross-Tabulation):
#### Investiguen la función pd.crosstab(). Generen una tabla cruzada entre la clase del pasajero (Pclass) y si sobrevivió (Survived). Esta técnica de sumarización matricial es el equivalente a una Tabla Dinámica en Excel. ¿Qué patrón logran visualizar rápidamente al cruzar ambas variables?

```python
cross_tabulation = pd.crosstab(df["Pclass"], df["Survived"])

print(cross_tabulation)
```

##### Que de la primera clase, la mayoría de las personas sobrevivieron, segunda clase fueron más quienes murieron pero los porcentajes están cercanos a un 50/50, mientras que en la tercera clase si fue fueron bastantes las personas que murieron siendo estos muchisimos más de la mitad de esta respectiva clase.
